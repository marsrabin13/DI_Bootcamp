### Identifying Interesting Columns for Analysis (Exercise 5)

Based on the column names, the following columns could be particularly interesting for analysis:

1.  **`Year` and `Avg hrs per day sleeping`**: These two columns are ideal for **trend analysis**. We can observe how the average hours of sleep per day have changed over the years, identifying any increasing or decreasing patterns.

2.  **`Age Group` and `Avg hrs per day sleeping`**: This allows for **group comparison** to understand if different age demographics have varying sleep patterns. This can reveal which age groups get more or less sleep on average.

3.  **`Sex` and `Avg hrs per day sleeping`**: Similar to `Age Group`, this enables **group comparison** to see if there are significant differences in sleep duration between men and women.

4.  **`Type of Days` and `Avg hrs per day sleeping`**: This column could be used for group comparison to analyze sleep patterns on different types of days (e.g., weekdays vs. weekends, holiday vs. non-holiday).

Let's start by looking at the unique values and distributions of these categorical columns to get a better understanding.

```python
print('Unique values in Age Group:', df['Age Group'].unique())
print('Unique values in Sex:', df['Sex'].unique())
print('Unique values in Type of Days:', df['Type of Days'].unique())
print('Unique values in Period:', df['Period'].unique())
```

Now let's visualize the trends and group comparisons for `Avg hrs per day sleeping`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter for 'Annual' period to get overall yearly trends if 'Period' column exists
if 'Period' in df.columns and 'Annual' in df['Period'].unique():
    df_annual = df[df['Period'] == 'Annual']
else:
    df_annual = df.copy()

# Trend analysis: Avg hrs per day sleeping over the years
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_annual, x='Year', y='Avg hrs per day sleeping', marker='o')
plt.title('Average Hours Per Day Sleeping Over Years (Annual Data)')
plt.xlabel('Year')
plt.ylabel('Average Hours Per Day Sleeping')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group comparison: Avg hrs per day sleeping by Age Group
plt.figure(figsize=(12, 7))
sns.barplot(data=df, x='Age Group', y='Avg hrs per day sleeping', ci=None, palette='viridis')
plt.title('Average Hours Per Day Sleeping by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Average Hours Per Day Sleeping')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group comparison: Avg hrs per day sleeping by Sex
plt.figure(figsize=(8, 6))
sns.barplot(data=df, x='Sex', y='Avg hrs per day sleeping', ci=None, palette='mako')
plt.title('Average Hours Per Day Sleeping by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Hours Per Day Sleeping')
plt.grid(axis='y')
plt.show()
```

🌟 Exercise 9 : Export a dataframe to excel format and JSON format.
Create a simple dataframe.
Export the dataframe to an excel file.
Export the dataframe to a JSON file.


```python
import pandas as pd

# Create a simple DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df_export = pd.DataFrame(data)

display(df_export)
print("DataFrame created successfully.")
```

```python
# Export DataFrame to an Excel file
excel_file_path = 'sample_data.xlsx'
df_export.to_excel(excel_file_path, index=False)
print(f"DataFrame successfully exported to {excel_file_path}")
```

```python
# Export DataFrame to a JSON file
json_file_path = 'sample_data.json'
df_export.to_json(json_file_path, orient='records', indent=4)
print(f"DataFrame successfully exported to {json_file_path}")
```

### Exercise 10: Reading JSON Data

For this exercise, we will use a publicly available JSON dataset. We'll load it using pandas and display the first few rows.

```python
import pandas as pd

# Sample JSON dataset URL (e.g., from an open API or public repository)
# This example uses a dataset about cryptocurrency prices, modify if needed
json_url = 'https://api.coindesk.com/v1/bpi/currentprice.json'

try:
    # Read the JSON data directly into a pandas DataFrame
    df_json = pd.read_json(json_url)

    print(f"Successfully loaded JSON data from: {json_url}")
    print("First five entries of the JSON data:")
    display(df_json.head())

except Exception as e:
    print(f"Error loading JSON data: {e}")
    print("Please check the URL or your internet connection.")
```
