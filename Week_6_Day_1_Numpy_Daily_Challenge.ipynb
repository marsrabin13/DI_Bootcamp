3. Data Visualization:



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(100)
temperatures = np.random.uniform(low=-5, high=35, size=(10, 12))
# 10 cities, 12 months
# Temperatures range from -5 to 35 degrees Celsius
city_temps = np.random.uniform(-5, 35, (10, 12))

city_names = ['Tokyo', 'London', 'New York', 'Sydney', 'Cairo',
              'Moscow', 'Mumbai', 'Berlin', 'Rio', 'Toronto']
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

print(f"Generated temperature data: {city_temps.shape}")
print(f"  {city_temps.shape[0]} cities × {city_temps.shape[1]} months")
print(f"\nFirst 3 cities, first 6 months:")
print(city_temps[:3, :6].round(1))

# Annual average per city: .mean(axis=1)
# Hottest/coldest city: .argmax(), .argmin()
annual_avg = city_temps.mean(axis=1)
hottest_city = annual_avg.argmax()
coldest_city = annual_avg.argmin()

# pd.DataFrame(city_temps, index=city_names, columns=month_names)
df = pd.DataFrame(city_temps, index=city_names, columns=month_names)
print(df)

#Visualization using line chart
plt.figure(figsize=(12,7)) # Increased figure size for better legend visibility
# Iterate through each city's temperature data to plot a separate line
for i, city_data in enumerate(city_temps):
    plt.plot(month_names, city_data, marker='o', linewidth=1, label=city_names[i]) # Removed 'color=green' to allow default color cycling
plt.title('Monthly Temperature Trends per City', fontsize=14, fontweight='bold') # Corrected typo 'CIty' to 'City'
plt.xlabel('Month')
plt.ylabel('Temperature (C)')
plt.grid(True)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') # Add legend outside the plot
plt.tight_layout() # Adjust layout to prevent legend cut-off
plt.show()
```
