### Instructions

```python
import pandas as pd
import os

# Priority 1: Check the current folder as per user context
local_path = './loan_prediction.csv'
# Priority 2: Check the full Drive path
drive_path = '/content/drive/MyDrive/Colab Notebooks/loan_prediction.csv'

try:
    if os.path.exists(local_path):
        df = pd.read_csv(local_path)
        print(f"Successfully loaded data from: {local_path}")
    elif os.path.exists(drive_path):
        df = pd.read_csv(drive_path)
        print(f"Successfully loaded data from: {drive_path}")
    else:
        raise FileNotFoundError
except (FileNotFoundError, Exception):
    print("Local file not found. Loading from verified public repository...")
    url = 'https://raw.githubusercontent.com/dphi-official/Datasets/master/Loan_Data/loan_train.csv'
    df = pd.read_csv(url)

print(f"Dataset Shape: {df.shape}")
display(df.head())
print("\nColumn Data Types:")
print(df.dtypes)
print("\nMissing per column:")
display(df.isna().sum().sort_values(ascending=False))
```

```python
import pandas as pd

# Selected features based on the loaded dataset
selected_features = [
    "Credit_History",
    "ApplicantIncome",
    "CoapplicantIncome",
    "LoanAmount",
    "Education",
    "Property_Area",
    "Married"
]

# 1. Handling Missing Values
# Continuous: Median Imputation
df['LoanAmount'] = df['LoanAmount'].fillna(df['LoanAmount'].median())

# Categorical: Mode Imputation (index 0 gets the most frequent value)
df['Credit_History'] = df['Credit_History'].fillna(df['Credit_History'].mode()[0])
df['Married'] = df['Married'].fillna(df['Married'].mode()[0])

# 2. One-Hot Encoding for categorical features
# This converts 'Education', 'Property_Area', and 'Married' into binary columns
df_encoded = pd.get_dummies(df[selected_features], columns=['Education', 'Property_Area', 'Married'], drop_first=True)

print("Preprocessing Complete:")
print(f"Original columns: {selected_features}")
print(f"Encoded DataFrame shape: {df_encoded.shape}")
display(df_encoded.head())

print("\nSummary of strategy:")
print("- One-Hot Encoding applied to categorical columns.")
print("- Missing 'LoanAmount' filled with Median.")
print("- Missing 'Credit_History'/'Married' filled with Mode.")
```

### Your justification
I have selected **Credit_History, ApplicantIncome, CoapplicantIncome, LoanAmount, Education, Property_Area, and Married** as the most relevant features.

*   **Credit_History** is the strongest predictor as it reflects past repayment behavior.
*   **ApplicantIncome** and **CoapplicantIncome** determine the total household capacity to repay the debt.
*   **LoanAmount** is critical because higher debt-to-income ratios increase default risk.
*   **Education** and **Married** status often correlate with financial stability and earning potential.

I have excluded **Loan_ID** because it is a unique identifier with no predictive power, and **Unnamed: 0** as it is just an index.

To handle categorical features like `Gender` or `Property_Area`, I would use **One-Hot Encoding** to convert them into numerical format. For missing values, I would use **Mode Imputation** for categorical variables like `Credit_History` and **Median Imputation** for continuous variables like `LoanAmount` to avoid the influence of outliers.

```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

### Your answer
For this loan default prediction task, I would choose **Logistic Regression** and **Random Forest** models. Logistic Regression is suitable because it provides clear coefficients for risk factor interpretation, while Random Forest is robust to outliers and can handle the non-linear relationships often found in financial data.

My evaluation plan involves splitting the data into a **70/30 train-test split** and using **5-fold stratified cross-validation** to ensure stable performance estimates. I will prioritize **F1-Score** and **Recall** as my primary metrics, as the cost of missing a potential default (False Negative) is significantly higher than the cost of a manual review for a false alarm. I will select the final decision threshold by analyzing the **Precision-Recall curve** to balance institutional risk appetite.

To address the class imbalance (as defaults are typically rarer than successful repayments), I will use **SMOTE (Synthetic Minority Over-sampling Technique)** or adjust the **class_weight** parameter in the models to penalize misclassifications of the minority class. Finally, I will iterate on hyperparameters using a **Randomized Search** while ensuring all preprocessing steps are contained within a pipeline to avoid data leakage.
